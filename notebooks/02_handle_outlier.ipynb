### 1. Import libraries

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from sklearn.model_selection import train_test_split
```

### 2. Load data set

```python
RAW_DATA_PATH = "../data/raw/energy_data_set.csv"
```

```python
df = pd.read_csv(RAW_DATA_PATH)
df
```

```python
""" Seperate numerical features from df """
df_cp = df.copy()

df_num = df_cp.select_dtypes(include=[np.number])

# create numerical colomns
numerical_col = df_num.columns.tolist()

numerical_col
```

### 3. Detect Outlier

#### 3.1 Using IQR

```python
def outlier_detection_iqr(col):
    Q1 = np.percentile(df[col], 25)
    Q3 = np.percentile(df[col], 75)

    IQR = Q3 - Q1

    upper_bound = Q3 + 1.5 * IQR
    lower_bound = Q1 - 1.5 * IQR

    return (df[col] > upper_bound) | (df[col] < lower_bound)
```

```python
print("using IQR")

for i in numerical_col:
    n_outlier = outlier_detection_iqr(i).sum()
    pct_outlier = (n_outlier / len(df_cp)) * 100
    print(f"{i} has : {n_outlier} outliers, as precentage : {pct_outlier:.2f}")
```

#### 3.2 Using Z-Score

```python
def outlier_detection_zscore(col):
    upper_threshold = 3
    lower_threshold = -3
    mean = np.mean(df[col])
    std = np.std(df[col])

    z = (df[col] - mean) / std

    return (z > upper_threshold) | (z < lower_threshold)
```

```python
print("using Z-Score")

for i in numerical_col:
    n_outlier = outlier_detection_zscore(i).sum()
    pct_outlier = (n_outlier / len(df_cp)) * 100
    print(f"{i} has : {n_outlier} outliers, as precentage : {pct_outlier:.2f}")
```

### 4. Handle Outlier

```python
df_cp
```

```python
"""Splitting data set to apply outlier handler"""

# Chronological Split
split_index = int(len(df_cp) * 0.8)

X = df_cp.drop(columns=['Appliances'])
y = df_cp['Appliances']

X_train_raw = X.iloc[:split_index]
X_test_raw  = X.iloc[split_index:]
y_train_raw = y.iloc[:split_index]
y_test_raw  = y.iloc[split_index:]
```

```python
"""Apply Cliping method as treatment"""

# Clip features using only training bounds
def clip_features_safely(X_train, X_test, columns, lower_pct=0.01, upper_pct=0.99):
    X_train_clipped = X_train.copy()
    X_test_clipped = X_test.copy()
    
    for col in columns:
        lower = X_train[col].quantile(lower_pct)
        upper = X_train[col].quantile(upper_pct)
        
        X_train_clipped[col] = X_train[col].clip(lower=lower, upper=upper)
        X_test_clipped[col] = X_test[col].clip(lower=lower, upper=upper)
        
        print(f"{col}: lower = {lower:.2f}, upper = {upper:.2f}")
    
    return X_train_clipped, X_test_clipped

cols_to_clip = ['lights', 'Visibility', 'RH_5']
X_train, X_test = clip_features_safely(X_train_raw, X_test_raw, cols_to_clip)
```

```python
# Check how many values were actually changed by clipping
for col in ['lights', 'Visibility', 'RH_5']:
    original_max = X_train_raw[col].max()
    clipped_max = X_train[col].max()
    print(f"{col}: Original max = {original_max:.2f} → Clipped max = {clipped_max:.2f}")
```

### 5. Save Outlier handled data set

```python
# Combine train , test
X_full = pd.concat([X_train, X_test], axis=0)
y_full = pd.concat([y_train_raw, y_test_raw], axis=0)

# Create full dataframe
df_full = X_full.copy()
df_full['Appliances'] = y_full

# Sort by index to maintain time order
df_full = df_full.sort_index()

print("Full dataset shape:", df_full.shape)
df_full.head()
```

```python
# Create folder if not exists
os.makedirs("../data/processed", exist_ok=True)

# Save as CSV
df_full.to_csv("../data/processed/cleaned_outlier_data.csv", index=False)

print("Full dataset saved successfully at: data/processed/cleaned_outlier_data.csv")
```
